### Missing Value Check
The dataset contains 9,994 rows and 21 columns. After checking missing values, all columns have 0 missing values, so no missing value treatment is required.

```python
df.duplicated().sum()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)
```

```python
df.info()
```

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Order Month Name"] = df["Order Date"].dt.month_name()
df["Order Quarter"] = df["Order Date"].dt.quarter
df["Shipping Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
df["Profit Margin %"] = (df["Profit"] / df["Sales"]) * 100
```

```python
df.head()
```

```python
df.shape
```

```python
df[["Sales", "Quantity", "Discount", "Profit", "Shipping Days", "Profit Margin %"]].describe()
```

### Data Cleaning Summary

- The dataset contains 9,994 rows and 21 original columns.
- No missing values were found in any column.
- Date columns were converted from text format to datetime format.
- New columns were created for year, month, quarter, shipping days, and profit margin.
- Sales and quantity values are valid because there are no zero or negative values.
- Profit contains negative values, which indicates loss-making orders.
- Discount values range from 0% to 80%, which is valid for business analysis.
- Shipping days range from 0 to 7 days, so there are no invalid shipping delays.

## Sales KPI Analysis
In this section, we calculate the main business KPIs to understand overall sales performance.

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_quantity = df["Quantity"].sum()
total_customers = df["Customer ID"].nunique()
average_order_value = total_sales / total_orders
profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Quantity Sold:", total_quantity)
print("Total Customers:", total_customers)
print("Average Order Value:", round(average_order_value, 2))
print("Profit Margin %:", round(profit_margin, 2))
```

### KPI Interpretation
These KPIs give a high-level view of business performance. Total sales shows revenue, total profit shows earnings, total orders shows customer demand, and profit margin shows overall profitability.

### KPI Findings

- Total Sales: 2,297,200.86
- Total Profit: 286,397.02
- Total Orders: 5,009
- Total Quantity Sold: 37,873
- Total Customers: 793
- Average Order Value: 458.61
- Profit Margin: 12.47%

The business generated total sales of 2.29M and total profit of 286K. The overall profit margin is 12.47%, which shows that the business is profitable, but there may still be opportunities to improve profit by analyzing discounts, products, and regions.

## Sales Trend Analysis

This section analyzes how sales changed over time using order dates. It helps identify yearly and monthly revenue patterns.

```python
yearly_sales = df.groupby("Order Year")["Sales"].sum().reset_index()

yearly_sales
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=yearly_sales, x="Order Year", y="Sales")
plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.show()
```

```python
monthly_sales = df.groupby(["Order Year", "Order Month"])["Sales"].sum().reset_index()

monthly_sales.head()
```

```python
monthly_sales["Year-Month"] = monthly_sales["Order Year"].astype(str) + "-" + monthly_sales["Order Month"].astype(str).str.zfill(2)

monthly_sales.head()
```

```python
plt.figure(figsize=(14,5))
sns.lineplot(data=monthly_sales, x="Year-Month", y="Sales", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.show()
```

```python
highest_sales_month = monthly_sales.loc[monthly_sales["Sales"].idxmax()]
lowest_sales_month = monthly_sales.loc[monthly_sales["Sales"].idxmin()]

print("Highest Sales Month:")
print(highest_sales_month)

print("\nLowest Sales Month:")
print(lowest_sales_month)
```

### Sales Trend Findings

The highest sales were recorded in November 2017, with total sales of 118,447.83. The lowest sales were recorded in February 2014, with total sales of 4,519.89.

The monthly trend shows that sales increased over time, especially toward the later years. November 2017 was the strongest sales month, which may indicate seasonal demand, year-end purchasing, or successful business activity during that period.

## Regional Sales Analysis

This section analyzes sales and profit performance across different regions to identify strong and weak markets.

```python
region_sales = df.groupby("Region")[["Sales", "Profit"]].sum().reset_index()
region_sales = region_sales.sort_values(by="Sales", ascending=False)

region_sales
```

```python
region_sales["Profit Margin %"] = (region_sales["Profit"] / region_sales["Sales"]) * 100
region_sales
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=region_sales, x="Region", y="Sales")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=region_sales, x="Region", y="Profit")
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.show()
```

```python
best_sales_region = region_sales.loc[region_sales["Sales"].idxmax()]
best_profit_region = region_sales.loc[region_sales["Profit"].idxmax()]
lowest_profit_region = region_sales.loc[region_sales["Profit"].idxmin()]

print("Best Region by Sales:")
print(best_sales_region)

print("\nBest Region by Profit:")
print(best_profit_region)

print("\nLowest Region by Profit:")
print(lowest_profit_region)
```

### Regional Sales Findings

The West region is the best-performing region with total sales of 725,457.82 and profit of 108,418.45. It also has a strong profit margin of 14.94%.

The Central region has sales of 501,239.89 but only 39,706.36 profit, with the lowest profit margin of 7.92%. This shows that Central is generating revenue, but profitability is weak compared to other regions.

Business Recommendation:
The company should continue investing in the West region because it performs strongly in both sales and profit. For the Central region, the company should review discount strategy, product mix, and operational costs to improve profitability.

### Product Performance Findings

Copiers are the most profitable sub-category, generating 55,617.82 profit with a strong profit margin of 37.20%. Phones and Accessories also contribute high profit, making them important product groups for the business.

Paper has a very high profit margin of 43.39%, even though its total sales are lower than Phones and Binders. This means Paper is a highly efficient profit-generating sub-category.

Tables are the biggest loss-making sub-category, with sales of 206,965.53 but a loss of 17,725.48. This shows that high sales do not always mean good business performance. Bookcases and Supplies also generated losses.

Business Recommendation:
The company should promote high-profit sub-categories such as Copiers, Phones, Accessories, and Paper. For loss-making sub-categories like Tables and Bookcases, the company should review discount levels, pricing strategy, and supplier costs.

```python
category_discount = df.groupby(["Category", "Discount"])[["Sales", "Profit"]].sum().reset_index()
category_discount["Profit Margin %"] = (category_discount["Profit"] / category_discount["Sales"]) * 100

category_discount.sort_values(by="Profit").head(10)
```

### Discount vs Profit Findings

The analysis shows that high discount levels are strongly linked with negative profit. Office Supplies at 80% discount generated sales of 16,963.76 but caused a loss of 30,539.04, resulting in a profit margin of -180.03%.

Technology at 70% discount and Office Supplies at 70% discount also caused major losses. Furniture shows repeated losses at 30%, 40%, 50%, 60%, and 70% discount levels.

This indicates that aggressive discounting may increase sales volume but can seriously reduce profitability.

Business Recommendation:
The company should review and limit high discount levels, especially discounts above 30%. Discount strategies should be monitored by category because some categories become highly unprofitable when large discounts are applied.
